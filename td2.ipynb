```python
import json
with open( "data/sample/instruments.json", encoding="utf-8") as file:
    instruments = json.load(file)
```

```python
print(type(instruments))
print(instruments.keys())
```

```python
instrument = instruments["instrument"]
benchmark = instruments["benchmark"]

print(instrument["ticker"])
print(instrument["name"])
print(instrument["currency"])

print(benchmark["ticker"])
print(benchmark["name"])
```

```python
import csv
def load_prices():
    with open("data/sample/prices.csv", encoding="utf-8") as file:
        return list(csv.DictReader(file))
```

```python
prices = load_prices()

print(type(prices))
print(type(prices[0]))
print(prices[0])
```

```python
print(prices[0]["close"])
print(type(prices[0]["close"]))
```

```python
close = float(prices[0]["close"])
```

```python
def filter_prices(prices, ticker):
    return [row for row in prices if row["ticker"] == ticker]
```

```python
instrument_prices = filter_prices(
    prices,
    instrument["ticker"],
)

benchmark_prices = filter_prices(
    prices,
    benchmark["ticker"],
)
```

```python
print(len(instrument_prices))
print(len(benchmark_prices))
```

```python
ticker = "AAPL"

result = []

for row in prices:
    if row["ticker"] == ticker:
        result.append(row)
```

```python
print(len(result))
```

```python
def get_first_close(prices):
    first_observation = prices[0]
    close = first_observation["close"]
    return float(close)
```

```python
print(get_first_close(instrument_prices))
print(get_first_close(benchmark_prices))
```

```python
def get_last_close(prices):
    last_observation = prices[-1]
    close = last_observation["close"]
    return float(close)
```

```python
print(get_last_close(instrument_prices))
print(get_last_close(benchmark_prices))
```

```python
def display_market_summary(asset, prices, show_currency=True):
    print(f'{asset["ticker"]} - {asset["name"]}')
    print(f'Observations : {len(prices)}')

    if show_currency:
        print(f'First close  : {get_first_close(prices):.2f} {asset["currency"]}')
        print(f'Last close   : {get_last_close(prices):.2f} {asset["currency"]}')
    else:
        print(f'First close  : {get_first_close(prices):.2f}')
        print(f'Last close   : {get_last_close(prices):.2f}')
```

```python
display_market_summary(instrument, instrument_prices)

print()

display_market_summary(benchmark, benchmark_prices, show_currency=False)
```
